### Q1. Two NumPy arrays have shapes (3, 1) and (1, 3). Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result differ from adding two one-dimensional arrays?


```python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
```

### 1. Predict the output of the given code.

  Result

[[11 12 13]
 [21 22 23]
 [31 32 33]]

Shape: (3, 3)

### 2. Identify the shapes of arrays a and b before addition.

Array a: Shape is (3, 1) (2D array: 3 rows, 1 column)

Array b: Shape is (1, 3) (2D array: 1 row, 3 columns)

### 3. Explain how NumPy broadcasting enables the addition of these arrays.

  NumPy compares array dimensions element-by-element starting from the trailing (rightmost) dimensions:
  1. Dimension 1 (Columns): a has size 1, and b has size 3. Since one of the dimensions is 1, a is stretched (replicated) 3 times horizontally across columns.
  
  2. Dimension 0 (Rows): a has size 3, and b has size 1. Since b has size 1, it is stretched (replicated) 3 times vertically across rows.

   Both arrays expand to an effective common shape of (3, 3) without physically copying memory:
   a behaves like: [[10 10 10
                     20 20 20
                     30 30 30]]
   b behaves like: [[1 2 3
                     1 2 3
                     1 2 3]]

   Adding them element-wise produces the final (3, 3) matrix.

### 4. What would happen if b were replaced with np.array([1, 2])? Explain the reason.
 
  If b is replaced with np.array([1, 2]), the code executes successfully without any error, producing a 2D array of shape (3, 2):
   [[11 12]
    [21 22]
    [31 32]]
    Shape: (3, 2)

Alignment: NumPy aligns array shapes starting from the rightmost (trailing) axis:
 a shape: (3, 1)
 b shape: (2,) automatically treated as (1, 2)

Broadcasting Rule: Dimensions are compatible if they are equal or if one of them is 1.
 Axis 1 (Columns): Size 1 vs 2,  a expands horizontally to size 2.
 Axis 0 (Rows): Size 3 vs 1 , b expands vertically to size 3.
 
Result: Both arrays expand to an effective shape of (3, 2), making the addition valid.

### Q1. A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing. 

### 1. Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().

```python
import pandas as pd

df = pd.read_csv("student_assessment_dirty.csv")

display(df.head())
display(df.info())
display(df.isnull().sum())
```

### Explanation:

head() displays the first five rows of the dataset.

info() shows the column names, data types, and non-null values.

isnull().sum() counts the missing values in each column.

### 2. Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

```python
df["Department"] = df["Department"].str.strip().str.lower()

df["Department"] = df["Department"].replace({
    "it": "IT",
    "i.t.": "IT",
    "finance": "Finance"
})

display(df["Department"].value_counts())
```

### Explanation:

str.strip() removes extra spaces.

str.lower() converts department names to lowercase so different versions can be matched.

replace() standardizes the department names.

value_counts() displays the number of records in each department.

### 3. Identify and remove exact duplicate student assessment records.

```python
original_count = len(df)
duplicates = df.duplicated().sum()
print("Number of duplicate records:", duplicates)

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

### Explanation:

duplicated().sum() counts the exact duplicate rows in the DataFrame.

drop_duplicates() removes duplicate rows while keeping the first occurrence.

shape displays the number of rows and columns remaining after removing duplicates.

### 4. Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.

```python
df["Marks"] = df["Marks"].astype(str).str.replace(" marks", "", regex=False)
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

display(df["Marks"])
display(df["Marks"].isnull().sum())
```

### Explanation:

astype(str) converts the values to strings so text can be cleaned.

str.replace() removes " marks" from values such as "88 marks".

pd.to_numeric(..., errors="coerce") converts valid values into numbers and changes values that cannot be converted, such as "Absent" and "NA", into NaN.

isnull().sum() counts the missing values after conversion.

### 5. Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

```python
missing_marks = df[df["Marks"].isnull()]
display(missing_marks)
print("Total missing marks:", df["Marks"].isnull().sum())
```

### Explanation:

isnull() identifies records where marks are missing.

Review the other columns in the displayed records to distinguish students marked as Absent from students whose marks have not been entered.

Absent students did not attend the assessment, while missing entries may indicate that their marks have not yet been recorded.

Risk of deleting missing values: If you delete all rows with missing marks without investigating, you may remove students whose marks simply have not been entered, resulting in loss of useful data.

### 6. Identify marks outside the valid range of 0–100 and decide how they should be handled.

```python
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

display(invalid_marks)
print("Number of invalid marks:", len(invalid_marks))
```

### Explanation:

df["Marks"] < 0 finds marks below 0.

df["Marks"] > 100 finds marks above 100.

| means OR, so the condition finds records matching either case.

len(invalid_marks) counts the invalid records.

Decision: Marks should normally be between 0 and 100. Review the invalid records and decide whether to correct them using reliable information or mark them as missing. Do not change values based on guesses.

### 7. Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

```python
display(df)

print("Records removed:", original_count - len(df))
print("Missing marks remaining:", df["Marks"].isnull().sum())
print("Invalid marks identified:", len(invalid_marks))
```

### Explanation:

display(df) shows the cleaned DataFrame.

original_count - len(df) calculates how many records were removed. This assumes duplicate removal was the only operation that removed rows.

isnull().sum() counts missing marks remaining after cleaning.

len(invalid_marks) reports how many invalid-mark records you identified in Step 6.
